```python
import numpy as np
import pandas as pd
from datetime import datetime
```

```python
now = datetime.now()

current_year = now.year
current_month = now.month
current_day = now.day
```

```python
df["car_age_months"] = (current_year - df["Year"]) * 12 + (current_month - 1)
```

```python
print(df[["Make", "Model", "Year", "car_age_months"]].head(10))
```
